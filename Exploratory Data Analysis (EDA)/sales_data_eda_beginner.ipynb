## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)
```

## 2. Load the dataset

Keep `sales_data.csv` in the same folder as this notebook. If you are using Google Colab, upload the CSV first.

```python
df = pd.read_csv('sales_data.csv')
df.head()
```

## 3. Understand the data

Check the dataset size, column names, data types, and basic numeric statistics.

```python
print('Rows and columns:', df.shape)
print('\nColumn names:')
print(df.columns.tolist())

df.info()
```

```python
df.describe()
```

## 4. Data quality checks

Find missing values and duplicate rows before analysis. Do not delete data without checking whether it is appropriate.

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Count duplicate rows
print('Duplicate rows:', df.duplicated().sum())

# Show rows where Payment_Method is missing
df[df['Payment_Method'].isnull()].head()
```

### Handle missing payment methods

The dataset has some missing `Payment_Method` values. For this beginner project, keep the rows and label the missing category as `Unknown` so those transactions are not lost.

```python
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
print('Missing values after handling:')
print(df.isnull().sum())
```

## 5. Prepare date column

Convert `Date` into a proper datetime value so we can group transactions by month.

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
print('Date range:', df['Date'].min(), 'to', df['Date'].max())
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df[['Date', 'Month']].head()
```

## 6. Key summary numbers

We use `Final_Amount` as the sales amount after discount. `Total_Amount` is before discount in this dataset.

```python

print('Number of transactions:', len(df))
print('Total units sold:', df['Units_Sold'].sum())
print('Sales before discount: ₹{:,.2f}'.format(df['Total_Amount'].sum()))
print('Sales after discount:  ₹{:,.2f}'.format(df['Final_Amount'].sum()))
print('Average final amount per transaction: ₹{:,.2f}'.format(df['Final_Amount'].mean()))
print('Average customer rating:', round(df['Customer_Rating'].mean(), 2))
```

## 7. Department-wise sales

Compare sales after discount by department.

```python
department_sales = df.groupby('Department')['Final_Amount'].sum().sort_values(ascending=False)
display(department_sales)

department_sales.plot(kind='bar', color='steelblue')
plt.title('Sales After Discount by Department')
plt.xlabel('Department')
plt.ylabel('Final Amount (₹)')
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()
```

## 8. Transactions by region

This chart counts transactions in each region. It does not show the amount of sales.

```python
region_counts = df['Region'].value_counts()
display(region_counts)

region_counts.plot(kind='bar', color='darkseagreen')
plt.title('Number of Transactions by Region')
plt.xlabel('Region')
plt.ylabel('Transaction Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 9. Payment method analysis

Because missing payment values were labeled `Unknown`, they will appear as a separate category.

```python
payment_counts = df['Payment_Method'].value_counts()
display(payment_counts)

payment_counts.plot(kind='bar', color='cornflowerblue')
plt.title('Transactions by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Transaction Count')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

## 10. Discount analysis

See how frequently different discount percentages were used.

```python
discount_counts = df['Discount_Percent'].value_counts().sort_index()
display(discount_counts)

sns.countplot(data=df, x='Discount_Percent')
plt.title('Transactions by Discount Percentage')
plt.xlabel('Discount (%)')
plt.ylabel('Transaction Count')
plt.tight_layout()
plt.show()
```

## 11. Customer ratings

Check the distribution of customer ratings from the dataset.

```python
rating_counts = df['Customer_Rating'].value_counts().sort_index()
display(rating_counts)

sns.countplot(data=df, x='Customer_Rating')
plt.title('Customer Rating Distribution')
plt.xlabel('Customer Rating')
plt.ylabel('Number of Transactions')
plt.tight_layout()
plt.show()
```

## 12. Monthly sales trend

Group final sales by month and plot the result. This works best when the dataset covers multiple months.

```python
monthly_sales = df.groupby('Month')['Final_Amount'].sum()
display(monthly_sales.head())

monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Sales After Discount')
plt.xlabel('Month')
plt.ylabel('Final Amount (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 13. Simple findings generated from the data

Run this cell to print a few factual observations. Check the results before writing your final conclusion.

```python
top_department = department_sales.idxmax()
top_region_by_transactions = region_counts.idxmax()
top_payment_method = payment_counts.idxmax()

print(f'Highest sales department: {top_department} (₹{department_sales.max():,.2f})')
print(f'Most transactions by region: {top_region_by_transactions} ({region_counts.max()} transactions)')
print(f'Most common payment method: {top_payment_method} ({payment_counts.max()} transactions)')
print(f'Average customer rating: {df["Customer_Rating"].mean():.2f} out of 5')
```
