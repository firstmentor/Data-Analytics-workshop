```python
import pandas as pd
```

```python
df = pd.read_csv("sales.csv")
df
```

```python
# Average salary department-wise:
```

```python
df.groupby("Department")["Monthly_Salary"].mean()
```

```python
# GroupBy + Sum
```

```python
df.groupby("Department")["Monthly_Salary"].sum()
```
