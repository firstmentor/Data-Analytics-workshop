```python
import pandas as pd
```

```python
data = {
    "Employee": ["Rahul", "Amit", "Neha", "Priya", "Rohit"],
    "Department": ["IT", "HR", "IT", "Sales", "IT"],
    "Salary": [40000, 30000, 50000, 35000, 45000],
    "Experience": [2, 1, 4, 2, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Salary"].mean() #Average salary
```

```python
df["Salary"].max()  #Highest salary?
```

```python
df[df["Department"] == "IT"]
```

```python
df.groupby("Department")["Salary"].mean()   #Department-wise average salary?
```

```python
# 3+ years experience wale employees?
df[df["Experience"] >= 3]
```
