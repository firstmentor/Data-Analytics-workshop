```python
import matplotlib.pyplot as plt
# pyplot Matplotlib ka ek module hai jo graph banane ke liye simple functions provide karta hai.
# matplotlib = poori library
# pyplot = graph banane wala module
# plt = short name (alias)
```

```python
# Line Chart
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 18000, 22000]

plt.plot(months, sales)

plt.show()
```

```python
plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# Line Chart kab use karein? 
# Line chart = Trend
# Example:

# Monthly sales
# Daily temperature
# Website visitors
# Revenue growth
```

```python
plt.figure(figsize=(4,3)) # figure size  inch
plt.plot(months, sales, color='y', marker='o', linestyle='--', linewidth=2, markersize=10)
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [10, 20, 25, 35, 40]
y2 = [20, 30, 35, 45, 55]

plt.plot(x, y1, label='Sales 2022') # ploting y1 data
plt.plot(x, y2, label='Sales 2023') # ploting y2 data

plt.title("PN Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend() # show legends
plt.grid(True)

plt.show()
```

```python
# '-'   Solid Line
# '--'  Dashed Line
# ':'   Dotted Line
# '-.'  Dash Dot

# 'o' Circle
# 's' Square
# '^' Triangle
# '*' Star
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Month':[1,2,3,4,5],
    'Sales':[10,20,15,5,10]
})

plt.plot(df['Month'],df['Sales'])

plt.show()
```
