```python
import matplotlib.pyplot as plt
```

```python
salary = [25000, 30000, 32000, 35000, 40000, 45000, 50000, 150000]
#8 employees ki salaries

plt.boxplot(salary)

plt.title("Salary Distribution")

plt.ylabel("Salary")

plt.show()
```

```python
# “Boxplot hume data ka spread, median aur outliers dikhata hai. 
# Is salary data mein ₹1,50,000 baaki salaries ke comparison mein bahut zyada hai, 
# isliye ye ek outlier hai.”
```

```python
# ₹25,000 se ₹50,000 tak salaries relatively close hain, lekin ₹1,50,000 bahut zyada hai.

# Isi ko Boxplot identify karne mein help karta hai.
# plt.boxplot(salary)
# Ye line Python ko bol rahi hai:

# "Salary ke data ka Boxplot bana do."

# Boxplot generally hume ye cheezein samajhne mein help karta hai:

# Data ka spread
# Median
# Lower/upper range
# Outliers

# Aapke graph mein jo orange horizontal line hai, wo median hai.
```

```python
# 3. Median kya hai?

# Aapka data sorted hai:
# 25,000
# 30,000
# 32,000
# 35,000
# 40,000
# 45,000
# 50,000
# 1,50,000
# Total 8 values hain.

# Middle ki 2 values:
# 35,000
# 40,000
# Median
# (35,000 + 40,000) / 2
# = 37,500
# Isliye box ke andar orange line approximately:
# ₹37,500

# par hai.

# Student ko bolo:

# "Median data ki middle value ko represent karta hai."
# 4. Box kya represent karta hai?

# Graph mein jo rectangle hai:

#        ┌──────────────┐
#        │              │
#        │      ─────   │ ← Median
#        │              │
#        └──────────────┘

# Ye middle 50% data ko represent karta hai.

# Is box ke:

# Bottom = Q1
# Orange line = Median
# Top = Q3

# Simple language mein:

# Box data ke beech ke 50% values ka spread dikhata hai.
```

```python
# 5. Whiskers kya hain?

# Box ke upar aur neeche jo lines ja rahi hain unhe whiskers bolte hain.

# Example:

#           ─────
#             │
#             │  ← Whisker
#        ┌──────────┐
#        │   ───    │
#        │          │
#        └──────────┘
#             │
#             │  ← Whisker
#           ─────

# Whiskers normal range ke data ko show karte hain.
```

```python
# 6. ₹1,50,000 kya hai?

# Ab sabse important point.

# Aapka data mostly:

# 25,000
# 30,000
# 32,000
# 35,000
# 40,000
# 45,000
# 50,000

# ke around hai.

# Lekin:

# 150,000

# bahut door hai.

# Isliye Boxplot ise outlier ke roop mein show karta hai.

# Graph mein ye usually ek alag dot ke form mein dikhega.

#         •  ← ₹1,50,000 OUTLIER
#         |
#       ──
#       │
#  ┌─────────┐
#  │   ───   │
#  └─────────┘
#       │
#      ───
# Student ko bolo:

# "₹1,50,000 baaki salaries se bahut zyada different hai, isliye Boxplot ise outlier ke roop mein identify 
# karta hai."
```

```python
# Ab poore graph ko ek story ki tarah samjho

# Maan lo ek company mein 8 employees hain.

# Unki salaries hain:

# ₹25k, ₹30k, ₹32k, ₹35k, ₹40k, ₹45k, ₹50k aur ₹1.5 lakh.

# Boxplot hume batata hai:

# 1. Orange line → Median
# Approximately ₹37,500

# 2. Box → Middle 50% data ka spread

# 3. Whiskers → Normal range

# 4. ₹1.5 lakh → Outlier

# Workshop mein ekdum simple line:

# "Boxplot salary data ka spread dikhata hai. Isme box middle 50% salaries ko show karta hai, 
# orange line median ko show karti hai, aur ₹1,50,000 ek outlier hai kyunki ye baaki salaries se 
# bahut zyada different hai."

# Important: Boxplot mein ₹1,50,000 ko outlier isliye bola jata hai kyunki wo IQR-based upper 
# limit se bahar hai, sirf isliye nahi ki wo visually bada number hai
```
