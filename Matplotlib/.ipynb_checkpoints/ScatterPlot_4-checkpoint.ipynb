```python
import matplotlib.pyplot as plt
```

```python
# Do numerical variables ke beech relationship dekhne ke liye.
```

```python
hours = [1, 2, 3, 4, 5] # Padhai ke ghante ($X$-axis par).
marks = [20, 40, 50, 70, 10]  #Students ke aaye hue marks ($Y$-axis par)

plt.scatter(hours, marks)
plt.title("Hours Studied vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")


plt.show()
```

```python
plt.scatter(
    hours,
    marks,
    color='red'
)
```

```python
plt.scatter(
    hours,
    marks,
    marker='*'
)
```

```python
# | Marker | Shape    |
# | ------ | -------- |
# | 'o'    | Circle   |
# | '*'    | Star     |
# | 's'    | Square   |
# | '^'    | Triangle |
# | 'x'    | Cross    |
```

```python
plt.scatter(
    hours,
    marks,
    s=200
)
```

```python
hours = [1,2,3,4,5]
marks = [20,40,50,70,90]

plt.scatter(
    hours,
    marks,
    color='blue',
    s=100
)

plt.title("Hours Studied vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid()

plt.show()
```

```python
# Data Analytics Interview Question

# Scatter Plot ka sabse common use:

# Study Hours vs Marks
# Height vs Weight
# Experience vs Salary
# Advertisement Cost vs Sales
# Temperature vs Ice Cream Sales
# Shortcut
# Scatter Plot

# Har Dot = Ek Record

# X-axis = Independent Variable
# Y-axis = Dependent Variable

# Use:
# Relationship / Correlation dikhane ke liye

# Ye Study Hours vs Marks wala example exam, viva aur interview me sabse zyada pucha jata hai.
```
