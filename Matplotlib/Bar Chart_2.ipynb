```python
import matplotlib.pyplot as plt
```

```python
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [50000, 80000, 30000, 20000]

plt.bar(products, sales,color="red") #color="red",width=0.5
# plt.grid(axis='y')   #Bar charts me mostly axis='y' use hota hai.
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python
# Categories compare karne ke liye:
```

```python
# Horizontal Bar Chart
plt.barh(products, sales)

plt.title("Product Sales")

plt.show()
```

```python

bars = plt.bar(products, sales)

plt.bar_label(bars,padding=3)   # Value automatically bar ke upar

plt.show()
```

```python
# Multiple Colors
colors = ['red','green','blue']
plt.figure(figsize=(8,5))
plt.bar(
    products,
    sales,
    color=colors
)

plt.show()
```
