```python
import numpy as np
```

```python
sales = np.array([12000, 15000, 18000, 10000, 22000])

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Highest Sales:", np.max(sales))
print("Lowest Sales:", np.min(sales))
print("Median Sales:", np.median(sales))
```

```python
# arange() Range ke andar numbers generate karne ke liye:
numbers = np.arange(1, 11)

print(numbers)
```

```python
numbers = np.arange(0, 21, 5)

print(numbers)
```

```python
# zeros()
arr = np.zeros(5)

print(arr)
```

```python
# ones()
arr = np.ones(5)

print(arr)
```

```python
# Random Numbers Ye 1–99 ke beech 5 random numbers generate karega.
numbers = np.random.randint(1, 100, 5)

print(numbers)
```

```python
# Reshape Array ka structure change kar sakte hain.
numbers = np.arange(1, 7)
# print(numbers)

data = numbers.reshape(2, 3)

print(data)
```
