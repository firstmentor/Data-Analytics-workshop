```python
# Array Type
```

```python
numbers = np.array([10, 20, 30, 40])

print(type(numbers))  #ndarray ka meaning N-dimensional array hai.
```

```python
# 1D Array
sales = np.array([1000, 2000, 3000, 4000])

print(sales)  #1-dimensional array
```

```python
# 2D Array Rows aur columns wale data ko 2D array mein represent kar sakte hain
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data)
```

```python
# Array Dimensions
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data.ndim)
```

```python
list1 =[1,2,3]
list2 =[4,5,6]
arr_list1_2 =np.array([list1,list2])
print(arr_list1_2)  #2d array axio
```

```python
# # axis 0 → Rows (ऊपर से नीचे)
# # axis 1 → Columns (बाएँ से दाएँ)
# 2 Rows
# 3 Columns
# arr[0][0] = 1
# arr[1][2] = 6
# पहला index = Row
# दूसरा index = Column
```

```python
list1 =[1,2,3]
list2 =[4,5,6]
list3 =[7,8,9]
arr_list1_3 =np.array([[list1,list2,list3]])
print(arr_list1_3)  #3d array axio
```

```python
# Shape
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data.shape)  #2 rows × 3 columns
```

```python
print(data.size) #Total elements:
```

```python
numbers = np.array([10, 20, 30])

print(numbers.dtype) #dtype batata hai array ke elements ka data type.
```
