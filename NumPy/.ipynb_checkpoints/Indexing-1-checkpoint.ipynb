```python
import numpy as np
```

```python
sales = np.array([1000, 2000, 3000, 4000])

print(sales[0])
print(sales[2])
```

```python
# Slicing
```

```python
sales = np.array([1000, 2000, 3000, 4000, 5000])

print(sales[1:4])
```

```python
# Mathematical Operations
```

```python
sales = np.array([1000, 2000, 3000, 4000])

print(sales + 100)
```

```python
print(sales * 2)
print(sales / 2)
```

```python
# Two Arrays par Operations
```

```python
sales1 = np.array([1000, 2000, 3000])
sales2 = np.array([500, 1000, 1500])

print(sales1 + sales2)
```

```python
# Important NumPy Functions
```

```python
sales = np.array([1000, 2000, 3000, 4000])

print(np.sum(sales))
```

```python
# Mean Average
print(np.mean(sales))
```

```python
# Maximum
print(np.max(sales))
```

```python
# Minimum
print(np.min(sales))
```

```python
# Standard Deviation
print(np.std(sales))
```

```python
# Median
print(np.median(sales))
```

```python
# # "Standard Deviation batata hai ki data ki values average ke around kitni faili hui hain."
# Low STD = values paas-paas
# High STD = values door-door
```

```python
# Data ki values average (mean) se kitni spread/faili hui hain.
```

```python
# Mean = Average
# Median = Middle value
# STD = Data kitna spread hai
```

```python
# # Median ka matlab hai sorted data ki beech wali value.
# # data = [10, 20, 30, 40, 50]
# # Beech wali value = 30
# # np.median(data)
# # Agar values even hon:
# data = [10, 20, 30, 40]
# Beech ki values = 20 aur 30

# Median:

# (20 + 30) / 2 = 25
```

```python
sales = np.array([12000, 15000, 18000, 10000, 22000])

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Highest Sales:", np.max(sales))
print("Lowest Sales:", np.min(sales))
print("Median Sales:", np.median(sales))
```
