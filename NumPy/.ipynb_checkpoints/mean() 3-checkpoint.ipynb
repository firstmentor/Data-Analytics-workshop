```python
import numpy as np
```

```python
# mean() — Average
sales = np.array([10, 20, 30, 40, 50])

print(np.mean(sales))
```

```python
# (10 + 20 + 30 + 40 + 50) / 5 = 30
# Matlab average sales 30 hai.
```

```python
# median() — Middle Value
# Median ka matlab beech ki value.
# Pehle data ko ascending order mein arrange karo:
# 10, 20, 30, 40, 50
```

```python
print(np.median(sales))
```

```python
data = np.array([10, 20, 30, 40])

print(np.median(data))
```

```python
# Beech ki 2 values:

# 20 aur 30

# Median:

# (20 + 30) / 2 = 25
```

```python
# std() — Standard Deviation
# Standard Deviation batata hai ki data values average se kitni spread/dispersed hain. 
# Spread ka matlab hai values ek doosre se kitni door hain, especially average se.
```

```python
data = np.array([48, 49, 50, 51, 52])

print(np.mean(data))
print(np.std(data))
# Values 50 ke aas-paas hain, isliye standard deviation low hoga.
```

```python
data = np.array([10, 30, 50, 70, 90])

print(np.mean(data))
print(np.std(data))
# Yahan values bahut spread hain, isliye standard deviation high hoga.
```

```python
# Simple rule:

# Low STD → Data values paas-paas hain

# High STD → Data values zyada spread hain
```

```python
# Ek real Data Analytics example
# Suppose 2 students ke marks hain:
# data = np.array([48, 49, 50, 51, 52])
# Dono ka mean = 50 hai.

# Lekin Group A ke marks 50 ke paas hain, jabki Group B ke marks bahut spread hain.

# Isliye:

# Mean same ho sakta hai, lekin Standard Deviation hume batata hai ki data kitna spread hai.

# Yaad rakhne ka easiest formula:

# Mean = Average

# Median = Middle

# STD = Spread
```

```python
# data = np.array([48, 49, 50, 51, 52])
# Step 1: Mean nikalo
# (48 + 49 + 50 + 51 + 52) / 5 = 50
# Mean = 50
# Ab dekho har value 50 se kitni door hai:
# | Value | 50 se distance |
# | ----: | -------------: |
# |    48 |              2 |
# |    49 |              1 |
# |    50 |              0 |
# |    51 |              1 |
# |    52 |              2 |
# Dekha? Sabhi values 50 ke bahut paas hain.

# Isliye hum bolte hain:

# Data ka spread low hai.
```

```python
# # data = np.array([10, 30, 50, 70, 90])
# Mean fir bhi 50 hai.
#     | Value | 50 se distance |
# | ----: | -------------: |
# |    10 |             40 |
# |    30 |             20 |
# |    50 |              0 |
# |    70 |             20 |
# |    90 |             40 |
# Yahan values 50 se kaafi door hain.

# Isliye:

# Data ka spread high hai
```

```python
# Low Spread

# 48  49  50  51  52
#         ↑
#        Mean


# High Spread

# 10       30       50       70       90
#                     ↑
#                    Mean
```

```python
# Spread ka matlab hai data ki values average ke aas-paas kitni faili hui hain. 
# Values average ke paas hain to spread low, 
# aur door-door hain to spread high."
```

```python
sales = np.array([12000, 15000, 18000, 10000, 22000])
```

```python
# 1. Total Sales

# 12,000 + 15,000 + 18,000 + 10,000 + 22,000 = ₹77,000

# 2. Average Sales

# ₹77,000 ÷ 5 = ₹15,400

# 3. Highest Sales

# ₹22,000

# 4. Lowest Sales

# ₹10,000

# 5. Median Sales

# Sorted data:

# 10,000, 12,000, 15,000, 18,000, 22,000
#                     ↑
#                  ₹15,000

# Median = ₹15,000

# 6. Standard Deviation

# ≈ ₹4,308.13
```

```python
import numpy as np

sales = np.array([12000, 15000, 18000, 10000, 22000])

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Highest Sales:", np.max(sales))
print("Lowest Sales:", np.min(sales))
print("Median Sales:", np.median(sales))
print("Standard Deviation:", np.std(sales))
```

```python
sales > np.mean(sales)
```

```python
# Find karo ki Average Sales se zyada sales wale days kitne hain. ye karo
```

```python
average = np.mean(sales)

above_average = sales > average

print(above_average)
print(np.sum(above_average))
```

```python
print(sales[sales > average])
```

```python
# 👉 Answer: ₹18,000 aur ₹22,000 — total 2 days.
```

```python
# # sales = np.array([12000, 15000, 18000, 10000, 22000])
# # Step 1: Mean nikalo
# # (12000 + 15000 + 18000 + 10000 + 22000) / 5
# # = 77000 / 5
# # = 15400

# # Mean = 15,400

# # Step 2: Har value se mean subtract karo
# |  Sales | Sales − Mean |
# | -----: | -----------: |
# | 12,000 |       -3,400 |
# | 15,000 |         -400 |
# | 18,000 |        2,600 |
# | 10,000 |       -5,400 |
# | 22,000 |        6,600 |
# # Step 3: Differences ka square karo
#     | Difference |     Square |
# | ---------: | ---------: |
# |     -3,400 | 11,560,000 |
# |       -400 |    160,000 |
# |      2,600 |  6,760,000 |
# |     -5,400 | 29,160,000 |
# |      6,600 | 43,560,000 |
# otal:

# 91,200,000

# # Step 4: 5 se divide karo
# 91,200,000 / 5
# = 18,240,000
# Step 5: Square root
# √18,240,000
# = 4270.83
# Correction: Maine pehle 4308.13 bataya tha, jo is dataset ke liye galat tha. np.std() ka correct result ≈ 4270.83 hai.
#     np.std(sales)
#     Simple definition students ko bolo:

# Standard Deviation batata hai ki data values mean se average तौर पर kitni spread hain.

# Yahan sales ka STD ₹4,270.83 hai.
```
