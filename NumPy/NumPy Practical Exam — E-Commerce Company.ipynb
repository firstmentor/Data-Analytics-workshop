```python
import numpy as np

sales = np.array([12500, 18000, 15000, 22000, 17500, 30000, 14000])
```

```python
# np.sum()
# np.mean()
# np.max()
# np.min()
# np.median()
# np.std()
# np.sort()
# sales > np.mean(sales)
```

```python
# Q1. Total Weekly Sales
print("Total Sales:", np.sum(sales))
```

```python
# Q2. Average Daily Sales
print("Average Sales:", np.mean(sales))
```

```python
# Q3. Highest & Lowest Sales
print("Highest Sales:", np.max(sales))
print("Lowest Sales:", np.min(sales))
```

```python
# Q4. Median Sales
print("Median Sales:", np.median(sales))
#12500, 14000, 15000, 17500, 18000, 22000, 30000
```

```python
# Q5. Standard Deviation
print("Standard Deviation:", np.std(sales))
#मतलब sales values average ke around approximately ₹5,435 tak spread ho rahi hain.
```

```python
# Q6. Average se zyada sales wale days
above_average = sales > np.mean(sales)

print(above_average)
print("Days:", np.sum(above_average))
```

```python
# Q7. Actual sales values
print(sales[sales > np.mean(sales)])
```

```python
# Q8. Ascending Order
print(np.sort(sales))
```

```python
# Q9. ₹20,000 se zyada sales
print(sales[sales > 20000])
```

```python
# # Q10. Business Insights
# Data se management ko ye insights milte hain:

# 1. Company ki weekly total sales ₹1,29,000 rahi.

# 2. Average daily sales ₹18,428.57 rahi.

# 3. Highest sales ₹30,000 rahi, jabki lowest sales ₹12,500 thi.

# 4. Sirf 2 days ki sales average se zyada rahi — ₹22,000 और ₹30,000.

# 5. Sales mein noticeable variation hai, kyunki Standard Deviation approximately ₹5,435 hai.
```

```python
import numpy as np

sales = np.array([12500, 18000, 15000, 22000, 17500, 30000, 14000])

print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Highest Sales:", np.max(sales))
print("Lowest Sales:", np.min(sales))
print("Median Sales:", np.median(sales))
print("Standard Deviation:", np.std(sales))

print("Above Average Sales:", sales[sales > np.mean(sales)])
print("Above Average Days:", np.sum(sales > np.mean(sales)))

print("Sorted Sales:", np.sort(sales))
print("Sales Above 20000:", sales[sales > 20000])
```
