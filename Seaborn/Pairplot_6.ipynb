```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [30000, 35000, 40000, 50000, 45000, 55000],
    "Experience": [1, 2, 2, 4, 3, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.pairplot(df)

plt.show()
```
