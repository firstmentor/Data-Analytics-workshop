```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python


data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [40000, 50000, 30000, 35000, 45000, 55000],
    "Experience": [2, 4, 1, 2, 3, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.barplot(data=df, x="Department", y="Salary")

plt.title("Department-wise Salary")

plt.show()
```

```python
# Category-wise comparison ke liye.
# Seaborn default mein har department ki average salary ka bar banata hai.
# Ye code IT, HR aur Sales departments ki average salary ko bar chart ke through compare karta hai."
```

```python
data = {
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing", "Furniture", "Furniture"],
    "Sales": [50000, 70000, 30000, 40000, 45000, 55000]
}

df = pd.DataFrame(data)

sns.barplot(data=df, x="Category", y="Sales")

plt.title("Category-wise Average Sales")
plt.xlabel("Category")
plt.ylabel("Average Sales")

plt.show()
```

```python
# | Category    |  Sales |
# | ----------- | -----: |
# | Electronics | 50,000 |
# | Electronics | 70,000 |
# | Clothing    | 30,000 |
# | Clothing    | 40,000 |
# | Furniture   | 45,000 |
# | Furniture   | 55,000 |
# Seaborn average nikalega:

# Electronics

# (50000 + 70000) / 2 = 60000

# Clothing

# (30000 + 40000) / 2 = 35000

# Furniture

# (45000 + 55000) / 2 = 50000

# To graph mein:

# Electronics → ₹60,000
# Furniture → ₹50,000
# Clothing → ₹35,000
# Business Insight:

# Electronics category ki average sales sabse zyada hai, jabki Clothing ki sabse kam hai.
```

```python
# Average ke bajay Total Sales ka barplot banana hai to pehle groupby().sum() karenge.

data = {
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing", "Furniture", "Furniture"],
    "Sales": [50000, 70000, 30000, 40000, 45000, 55000]
}

df = pd.DataFrame(data)

# Category-wise Total Sales
category_sales = df.groupby("Category", as_index=False)["Sales"].sum()

print(category_sales)

# Barplot
sns.barplot(data=category_sales, x="Category", y="Sales")

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Important: Pehle wale code mein barplot average nikal raha tha. Is code mein humne Pandas se total pehle hi nikal diya, phir Seaborn se visualize kiya.
```

```python
# True → Grouping column ko index bana do.
# False → Grouping column ko normal column rakho.
```
