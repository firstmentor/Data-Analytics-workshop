```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python


# Simple Data: Study Hours vs Marks
data = {"Study_Hours": [1, 2, 3, 4, 5], "Marks": [30, 45, 60, 80, 90]}

df = pd.DataFrame(data)

# 1. Correlation Matrix Nikalna
corr_matrix = df.corr()
print(corr_matrix)

# 2. Heatmap Banana
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Study Hours vs Marks Correlation")
plt.show()
```
