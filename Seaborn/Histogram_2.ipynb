```python
# "Histogram numerical data ko different ranges (bins) mein divide karta hai aur har range mein kitni values hain, ye count karke graph mein dikhata hai."
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
# Data distribution dekhne ke liye:
```

```python

data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
     "Salary": [30000, 35000, 40000, 50000, 45000, 55000],
     "Experience": [1, 2, 2, 4, 3, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.histplot(data=df, x="Salary")

plt.title("Salary Distribution")

plt.show()
```

```python
# "Mere paas 100 employees ki salary hai. Main dekhna chahta hoon ki salary kis range mein kitne employees hain. Is distribution ko dekhne ke liye hum Histogram use karte hain."
```

```python
# Aapke data mein:

# 30000
# 35000
# 40000
# 45000
# 50000
# 55000
# Step 1: Minimum aur maximum

# Sabse chhoti value:

# 30000

# Sabse badi:

# 55000

# To total range:

# 55000 - 30000 = 25000
# Step 2: Automatic bins

# Aapne:

# sns.histplot(data=df, x="Salary")

# likha hai.

# Aapne bins= nahi diya, isliye Seaborn automatic binning method use karta hai. Is particular data par 4 bins choose hue.

# Total range 25,000 ko 4 equal parts mein divide kiya:

# 25000 ÷ 4 = 6250

# Isliye boundaries approximately:

# 30,000
# 36,250
# 42,500
# 48,750
# 55,000

# Yaani actual bins roughly:

# 30,000 ─ 36,250
# 36,250 ─ 42,500
# 42,500 ─ 48,750
# 48,750 ─ 55,000
# Ab values ko dekho

# 1st bin: 30,000–36,250

# 30000 ✅
# 35000 ✅

# Count = 2

# 2nd bin: 36,250–42,500

# 40000 ✅

# Count = 1

# 3rd bin: 42,500–48,750

# 45000 ✅

# Count = 1

# 4th bin: 48,750–55,000

# 50000 ✅
# 55000 ✅

# Count = 2

# 🔥 Ek correction

# Graph dekhkar maine pehle 30,000–36,000 kaha tha. More accurately, yahan boundary 36,250 ke around hai, 36,000 nahi.

# Students ko exact formula yaad karane ki zarurat nahi hai. Bas ye samjhao:

# "Agar hum bins nahi dete, to Seaborn automatically data ke according bins ki boundaries decide karta hai. Phir har bin mein kitni values hain, uska count graph mein dikhata hai."
```

```python
sns.histplot(data=df, x="Salary", kde=True)

plt.title("Salary Distribution")

plt.show()
```
