```python
# Boxplot data ka spread, median, middle 50% data aur outliers ko dikhata hai.
# Boxplot hume data ki range, median, middle 50% values aur outliers ko easily samajhne mein help karta hai.”
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Salary": [40000, 50000, 30000, 35000, 45000, 55000],
    
}

df = pd.DataFrame(data)

print(df)
```

```python
# Boxplot
# Spread aur outliers dekhne ke liye:
sns.boxplot(data=df, y="Salary")

plt.title("Salary Distribution")

plt.show()
```

```python
# ### 1. Upar aur Niche ki Lines — Whiskers

# **Sabse Niche ki Horizontal Line (30,000):**
# Yeh data ki **minimum salary** hai, yani sabse kam salary **₹30,000** hai.

# **Sabse Upar ki Horizontal Line (55,000):**
# Yeh data ki **maximum salary** hai, yani sabse zyada salary **₹55,000** hai.

# ---

# ### 2. Blue Color ka Box — Interquartile Range (IQR)

# Yeh box data ke **middle 50% salaries** ki range ko dikhata hai.

# **Box ka Niche Edge — Q1 = ₹36,250:**
# Iska matlab hai ki **25% employees ki salary ₹36,250 ya usse kam** hai.

# **Box ke Beech ki Black Line — Median = ₹42,500:**
# Median data ka **middle value** hota hai.
# Iska matlab hai ki **50% employees ki salary ₹42,500 ya usse kam** hai aur **50% employees ki salary ₹42,500 ya usse zyada** hai.

# **Box ka Upar Edge — Q3 = ₹48,750:**
# Iska matlab hai ki **75% employees ki salary ₹48,750 ya usse kam** hai.

# Isliye **₹36,250 se ₹48,750** ke beech ki range mein data ka **middle 50%** hai.

# ---

# ### 3. Outliers — Extra Dots

# Agar boxplot mein box ke upar ya niche **alag se dots/points** dikhte hain, to unhe **Outliers** kaha jata hai.

# **Is graph mein koi extra dot nahi hai**, iska matlab hai ki **koi outlier detect nahi hua**.

# Yani is data mein koi salary value **normal range se unusually zyada ya kam** nahi hai.
```

```python
# Category ke according:
sns.boxplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.title("Department-wise Salary Distribution")

plt.show()
```
